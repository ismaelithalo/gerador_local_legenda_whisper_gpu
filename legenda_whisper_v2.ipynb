```python
# @title Inicializa bibliotecas
import whisper
import os
import sys
```

```python
# @title Define o modelo do Whisper que será utilziado

model = whisper.load_model("medium")
#medium +-rapido +-preciso
#large --rapido ++preciso
#turbo ++rapido --preciso
```

```python
# @title Função que gera a legenda em cima da transcrição do whisper

def format_time(seconds):
    hours, remainder = divmod(seconds, 3600)
    minutes, seconds = divmod(remainder, 60)
    millis = int((seconds - int(seconds)) * 1000)
    return f"{int(hours):02}:{int(minutes):02}:{int(seconds):02},{millis:03}"

def transcribe_to_srt(lang, filepath, filename):
    result = model.transcribe(os.path.join(filepath, filename), language=lang, task="translate")
    #Caso tenha separado o audio do video
    #result = model.transcribe("audio.mp3", language=lang, task="translate")

    srt_filename = os.path.splitext(filename)[0] + '.srt'

    with open(srt_filename, 'w') as srt_file:
        total_segments = len(result['segments'])

        for i, segment in enumerate(result['segments']):
            start = segment['start']
            end = segment['end']
            text = segment['text'].strip()

            progress = (i + 1) / total_segments * 100
            print(f"Progresso: {i + 1}/{total_segments} ({progress:.2f}%)")

            srt_file.write(f"{i+1}\n")
            srt_file.write(f"{format_time(start)} --> {format_time(end)}\n")
            srt_file.write(f"{text}\n\n")
```

```python
# @title [Alterar aqui] Configura os caminhos e arquivos que serão utilizados

default_root_drive_path = "/content/drive/MyDrive/"

#Caminho para a pasta do arquivo a partir da sua home do Google Drive
video_folder = "15 - Videos/YT/AVAM"

#Nome do arquivo cuja legenda será gerada
video_filename = "【生誕祭】東雲青空生誕祭2024裏側密着🌏💙 [3qBOhPeO50o].mp4"

#Idioma do video
video_language = "ja"
#ja ko pt en zh

video_full_path = default_root_drive_path + video_folder
sys.path.append(video_full_path)
```

```python
# @title Gera o arquivo de legenda
transcribe_to_srt(video_language, video_full_path, video_filename)
```
